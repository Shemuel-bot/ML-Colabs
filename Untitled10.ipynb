```python
# Comprehensive list of all official Sketch-RNN categories and mashups
sketch_rnn_categories = [
    "alarm_clock", "ambulance", "angel", "ant", "antyoga", "backpack", "barn",
    "basket", "bear", "bee", "beeflower", "bicycle", "bird", "book", "brain",
    "bridge", "bulldozer", "bus", "butterfly", "cactus", "calendar", "castle",
    "cat", "catbus", "catpig", "chair", "couch", "crab", "crabchair",
    "crabrabbitfacepig", "cruise_ship", "diving_board", "dog", "dogbunny",
    "dolphin", "duck", "elephant", "elephantpig", "eye", "face", "fan",
    "fire_hydrant", "firetruck", "flamingo", "flower", "floweryoga", "frog",
    "frogsofa", "garden", "hand", "hedgeberry", "hedgehog", "helicopter",
    "kangaroo", "key", "lantern", "lighthouse", "lion", "lionsheep", "lobster",
    "map", "mermaid", "monapassport", "monkey", "mosquito", "octopus", "owl",
    "paintbrush", "palm_tree", "parrot", "passport", "peas", "penguin", "pig",
    "pigsheep", "pineapple", "pool", "postcard", "power_outlet", "rabbit",
    "rabbitturtle", "radio", "radioface", "rain", "rhinoceros", "rifle",
    "roller_coaster", "sandwich", "scorpion", "sea_turtle", "sheep", "skull",
    "snail", "snowflake", "speedboat", "spider", "squirrel", "steak", "stove",
    "strawberry", "swan", "swing_set", "the_mona_lisa", "tiger", "toothbrush",
    "toothpaste", "tractor", "trombone", "truck", "whale", "windmill", "yoga", "yogabicycle"
]

def get_clip_embeddings_for_category(category_name):
    # Format category names cleanly (e.g., 'cruise_ship' -> 'cruise ship')
    clean_name = category_name.replace("_", " ")

    templates = [
        f"a minimalist line drawing of a {clean_name}",
        f"a quick sketch of a {clean_name}",
        f"hand-drawn doodle of a {clean_name}",
        f"a vector illustration of a {clean_name}"
    ]

    text_tokens = clip.tokenize(templates).to(device)
    with torch.no_grad():
        embeddings = clip_model.encode_text(text_tokens)
        mean_embedding = embeddings.mean(dim=0, keepdim=True)
    return mean_embedding.cpu().numpy()[0]
```

```python
import torch.nn.functional as F
class TextToSketchLatentDataset(Dataset):
    def __init__(self, clip_embeddings, latent_vectors):
        self.clip_embeddings = torch.tensor(clip_embeddings, dtype=torch.float32)
        self.latent_vectors = torch.tensor(latent_vectors, dtype=torch.float32)

    def __len__(self):
        return len(self.clip_embeddings)

    def __getitem__(self, idx):
        return self.clip_embeddings[idx], self.latent_vectors[idx]

# Build dataset across all categories
print("Generating text embeddings and mapping arrays for all categories...")
all_embeddings = []
all_latents = []

samples_per_category = 50  # Increase if pairing with actual .npz extracted latents

for cat in tqdm(sketch_rnn_categories):
    text_emb = get_clip_embeddings_for_category(cat)

    # In production, replace this mock array with real latents extracted
    # via the Sketch-RNN encoder for each category's .npz vector strokes.
    simulated_latents = np.random.randn(samples_per_category, 128)

    for z in simulated_latents:
        all_embeddings.append(text_emb)
        all_latents.append(z)

X_data = np.array(all_embeddings, dtype=np.float32)
y_data = np.array(all_latents, dtype=np.float32)

dataset = TextToSketchLatentDataset(X_data, y_data)
dataloader = DataLoader(dataset, batch_size=128, shuffle=True)
```

```python

# Define the Adapter Network
class CLIPToSketchRNNAdapter(nn.Module):
    def __init__(self, clip_dim=512, latent_dim=128, hidden_dim=512):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(clip_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(0.1),
            nn.Linear(hidden_dim, hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, latent_dim)
        )

    def forward(self, x):
        return self.net(x)

adapter = CLIPToSketchRNNAdapter().to(device)
optimizer = optim.AdamW(adapter.parameters(), lr=3e-4)
criterion = nn.MSELoss()


epochs = 100
print("Starting adapter training with accuracy tracking...")

for epoch in range(epochs):
    adapter.train()
    total_loss = 0
    total_cos_sim = 0

    for clip_emb, target_z in dataloader:
        clip_emb, target_z = clip_emb.to(device), target_z.to(device)

        optimizer.zero_grad()
        pred_z = adapter(clip_emb)
        loss = criterion(pred_z, target_z)

        loss.backward()
        optimizer.step()

        # Track loss
        total_loss += loss.item()

        # Calculate directional accuracy using Cosine Similarity (converted to a percentage)
        cos_sim = F.cosine_similarity(pred_z, target_z, dim=1).mean().item()
        total_cos_sim += cos_sim

    avg_loss = total_loss / len(dataloader)
    avg_accuracy = (total_cos_sim / len(dataloader)) * 100

    print(f"Epoch {epoch+1}/{epochs} | Loss: {avg_loss:.4f} | Alignment Accuracy: {avg_accuracy:.2f}%")

# Save weights after training
torch.save(adapter.state_dict(), "clip_sketch_adapter.pth")
print("Training complete! Model weights saved.")
```

```python
import json

# Extract weights and biases from the trained PyTorch adapter
weights_dict = {}
for name, param in adapter.named_parameters():
    weights_dict[name] = param.detach().cpu().numpy().tolist()

# Save as a JSON file
with open('clip_sketch_adapter_weights.json', 'w') as f:
    json.dump(weights_dict, f)

print("Weights exported to clip_sketch_adapter_weights.json!")
```
