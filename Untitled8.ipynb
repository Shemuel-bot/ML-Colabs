```python
# Cell 1: Setup
!pip install torch --quiet

import torch
import torch.nn as nn
import torch.nn.functional as F
import random
import math

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

```

```python
# Cell 2: Data generation and tokenization
# Task: "1+N=Answer" for N in [0, 200]

vocab = list("0123456789+=") + ["<pad>", "<eos>"]
stoi = {ch: i for i, ch in enumerate(vocab)}
itos = {i: ch for ch, i in stoi.items()}
PAD_ID = stoi["<pad>"]
EOS_ID = stoi["<eos>"]

def make_example(n):
    q = f"1+2+{n}="
    a = f"{1+2+n}"
    return q, a

def encode(s):
    return [stoi[c] for c in s]

# Build full dataset
data = [make_example(n) for n in range(20000)]  # 201 examples (0..200 inclusive)

# Pad sequences to a fixed length
max_q_len = max(len(q) for q, a in data)
max_a_len = max(len(a) for q, a in data)
max_len = max_q_len + max_a_len + 1  # +1 for eos

def encode_example(q, a):
    full = q + a
    ids = encode(full) + [EOS_ID]
    ids += [PAD_ID] * (max_len - len(ids))
    # mask: 1 where we want to compute loss (the answer + eos region), 0 elsewhere
    loss_mask = [0] * len(q) + [1] * (len(a) + 1) + [0] * (max_len - len(q) - len(a) - 1)
    return ids, loss_mask

encoded = [encode_example(q, a) for q, a in data]

# Random 50/50 interpolation split
indices = list(range(len(data)))
#random.shuffle(indices)
split = int(0.05 * len(indices))
train_idx, test_idx = indices[:split], indices[split:]

print(f"Train size: {len(train_idx)}, Test size: {len(test_idx)}")
print("Example:", data[train_idx[0]])
```

```python
# Cell 3: Batching helper

def get_batch(idx_list, batch_size=32):
    batch_idx = random.sample(idx_list, min(batch_size, len(idx_list)))
    xs, masks = zip(*[encoded[i] for i in batch_idx])
    x = torch.tensor(xs, dtype=torch.long)
    m = torch.tensor(masks, dtype=torch.long)
    return x.to(device), m.to(device)
```

```python
# Cell 4: Small GPT-style decoder-only transformer

class CausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_head):
        super().__init__()
        self.n_head = n_head
        self.d_head = d_model // n_head
        self.qkv = nn.Linear(d_model, 3 * d_model)
        print(self.qkv)
        self.proj = nn.Linear(d_model, d_model)

    def forward(self, x):
        B, T, C = x.shape
        qkv = self.qkv(x).view(B, T, 3, self.n_head, self.d_head).permute(2, 0, 3, 1, 4)
        q, k, v = qkv[0], qkv[1], qkv[2]  # (B, n_head, T, d_head)
        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.d_head)
        causal_mask = torch.tril(torch.ones(T, T, device=x.device)).bool()
        att = att.masked_fill(~causal_mask, float("-inf"))
        att = F.softmax(att, dim=-1)
        out = att @ v
        out = out.transpose(1, 2).contiguous().view(B, T, C)
        return self.proj(out)

class Block(nn.Module):
    def __init__(self, d_model, n_head):
        super().__init__()
        self.ln1 = nn.LayerNorm(d_model)
        self.attn = CausalSelfAttention(d_model, n_head)
        self.ln2 = nn.LayerNorm(d_model)
        self.mlp = nn.Sequential(
            nn.Linear(d_model, 4 * d_model),
            nn.GELU(),
            nn.Linear(4 * d_model, d_model),
        )

    def forward(self, x):
        x = x + self.attn(self.ln1(x))
        x = x + self.mlp(self.ln2(x))
        return x

class TinyGPT(nn.Module):
    def __init__(self, vocab_size, max_len, d_model=64, n_head=4, n_layer=3):
        super().__init__()
        self.tok_emb = nn.Embedding(vocab_size, d_model)
        self.pos_emb = nn.Embedding(max_len, d_model)
        self.blocks = nn.ModuleList([Block(d_model, n_head) for _ in range(n_layer)])
        self.ln_f = nn.LayerNorm(d_model)
        self.head = nn.Linear(d_model, vocab_size)

    def forward(self, idx):
        B, T = idx.shape
        pos = torch.arange(T, device=idx.device).unsqueeze(0)
        x = self.tok_emb(idx) + self.pos_emb(pos)
        for block in self.blocks:
            x = block(x)
        x = self.ln_f(x)
        return self.head(x)

model = TinyGPT(len(vocab), max_len).to(device)
print(f"Params: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# Cell 5: Training loop

optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4)
n_steps = 3600
batch_size = 32

for step in range(n_steps):
    x, mask = get_batch(train_idx, batch_size)
    logits = model(x[:, :-1])
    targets = x[:, 1:]
    loss_mask = mask[:, 1:]

    loss = F.cross_entropy(
        logits.reshape(-1, logits.size(-1)),
        targets.reshape(-1),
        reduction="none",
    )
    loss = (loss * loss_mask.reshape(-1)).sum() / loss_mask.sum().clamp(min=1)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if step % 300 == 0:
        print(f"step {step}: loss {loss.item():.4f}")
```

```python
# Cell 6: Evaluation — full accuracy + generation-based check
# We generate the answer autoregressively from "1+N=" rather than teacher-forcing,
# since that's the real test of generalization.

@torch.no_grad()
def generate_answer(n):
    q, a_true = make_example(n)
    ids = encode(q)
    x = torch.tensor([ids], dtype=torch.long, device=device)
    for _ in range(max_a_len + 1):
        logits = model(x)
        next_id = logits[0, -1].argmax().item()
        if next_id == EOS_ID:
            break
        x = torch.cat([x, torch.tensor([[next_id]], device=device)], dim=1)
    pred_ids = x[0, len(ids):].tolist()
    pred = "".join(itos[i] for i in pred_ids if i not in (PAD_ID, EOS_ID))
    return pred, a_true

def is_carry_case(n):
    # "1+N" requires a carry when N ends in 9 (units digit rolls over)
    return str(n)[-1] == "9"

def evaluate(idx_list, label):
    correct = 0
    carry_correct, carry_total = 0, 0
    nocarry_correct, nocarry_total = 0, 0
    wrong_examples = []

    for i in idx_list:
        n = int(data[i][0].split("+")[1].rstrip("="))
        pred, true = generate_answer(n)
        ok = pred == true
        correct += ok
        if is_carry_case(n):
            carry_total += 1
            carry_correct += ok
        else:
            nocarry_total += 1
            nocarry_correct += ok
        if not ok:
            wrong_examples.append((n, true, pred))

    acc = correct / len(idx_list)
    print(f"\n{label} accuracy: {acc:.2%} ({correct}/{len(idx_list)})")
    if carry_total:
        print(f"  Carry cases (N ending in 9): {carry_correct}/{carry_total} = {carry_correct/carry_total:.2%}")
    if nocarry_total:
        print(f"  Non-carry cases: {nocarry_correct}/{nocarry_total} = {nocarry_correct/nocarry_total:.2%}")
    if wrong_examples:
        print(f"  Sample errors (n, true, pred): {wrong_examples[:10]}")
    return acc

train_acc = evaluate(train_idx, "Train")
test_acc = evaluate(test_idx, "Test")
```
